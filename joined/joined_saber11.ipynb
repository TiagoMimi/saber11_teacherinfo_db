## Quality of the data 

```python
def dataframe_quality_function(dataframe):

    try:
        column_summary = []
        for col in dataframe.columns:                                      # *Iterate through the dataset columns*
            total_rows = len(dataframe)                                    # *Calculate the total number of rows*
            n_nulls = dataframe[col].isna().sum()                         # *Calculate the number of null values per column*
            percent_null = ((n_nulls / total_rows) * 100).round(2)        # *Calculate the percentage of null values*
            non_null = total_rows - n_nulls                               # *Calculate the number of non-null values*
            percent_non_null = ((non_null / total_rows) * 100).round(2)   # *Calculate the percentage of non-null values*
            dtype = dataframe[col].dtype                                   # *Get the data type of the column*
            column_summary.append({                                       # *Create the data quality summary for reporting or export*

                "Column_Name": col,
                "Data_Type": dtype,
                "Valid_Records": non_null,
                "Percentage_Valid_Records": percent_non_null,
                "Null_Records": n_nulls,
                "Percentage_Null_Records": percent_null
            })

        df_quality_summary = pd.DataFrame(column_summary)
        df_quality_summary.to_excel("saber11_dataframe_quality.xlsx")
        print("Excel completed")

        return df_quality_summary
    except Exception as e:
        print("Excel creation was not completed")
```

```python
dataframe_quality_function(df_saber11)
```

We will organize the information into three main categories:

- **Student Information:** Variables describing the student's personal, academic, and demographic characteristics.

- **School Information:** Variables related to the educational institution, such as school characteristics and location.

- **Socioeconomic Information:** Variables describing the student's family background, socioeconomic conditions, and household characteristics.

- **Student Results:** Variables related to the student's academic performance and results obtained in the Saber 11 examination.

## Write csv

```python
import json
```

```python
# Output folder for joined CSV files
joined_path = project_path / "data" / "joined"

# Create folder if it does not exist
joined_path.mkdir(parents=True, exist_ok=True)
```

```python
dataframes = {
    "student_info": df_student_info,
    "school_info": df_school_info,
    "result_info": df_result_info,
    "socioeconomic_info": df_socieconomic_info,
}

# Dictionary to store the data types of each DataFrame
data_types = {}

for name, df in dataframes.items():

    # Save DataFrame as CSV
    df.to_csv(
        joined_path / f"{name}.csv",
        sep=";",
        index=False,
        encoding="utf-8-sig"
    )

    # Save column data types
    data_types[name] = {
        column: str(dtype)
        for column, dtype in df.dtypes.items()
    }

# Save data types as JSON
with open(
    joined_path / "data_types_student.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        data_types,
        file,
        indent=4,
        ensure_ascii=False
    )
```
