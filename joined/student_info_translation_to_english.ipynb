```python
data_path = project_path / "saber11_db" / "data" / "joined" 

# Read student data types
with open(
    data_path / "data_types_student.json",
    "r",
    encoding="utf-8"
) as file:
    data_type_student = json.load(file)


# List of CSV files
file_list = [
    data_path / "student_info.csv"
]

# DataFrame names
df_names = [
    "df_student_info"]


# Read CSV files using the data types stored in the JSON files
dataframes = {}

for name, file in zip(df_names, file_list):

    # Get table name
    table_name = name.replace("df_", "")

    # Select the corresponding data type dictionary
    if table_name.startswith("teacher_"):
        dtype_mapping = data_type_teacher[table_name]
    else:
        dtype_mapping = data_type_student[table_name]

    # Read CSV
    dataframes[name] = pd.read_csv(
        file,
        sep=";",
        dtype=dtype_mapping,
        low_memory=False
    )


# Assign DataFrames
df_student_info = dataframes["df_student_info"]
```

### Fuction_to_translate

```python
def translate_values_dictionary(df,column,dict_translation):
    
    df[column] = df[column].replace(dict_translation)

    return df
```

#### estu_mcpio_reside	=> STUDENT_MUNICIPALITY_RESIDENCE

```python
df_student_info = df_student_info.rename(columns={"estu_mcpio_reside":"STUDENT_MUNICIPALITY_RESIDENCE"})
```

```python
import unicodedata

df_student_info["STUDENT_MUNICIPALITY_RESIDENCE"] = (
    df_student_info["STUDENT_MUNICIPALITY_RESIDENCE"]
    .apply(lambda x: ''.join(
        c for c in unicodedata.normalize('NFD', str(x))
        if unicodedata.category(c) != 'Mn'
    ).lower().replace(' ', '_'))
)
```
